```python
target_table = f"{catalogo}.{esquema_gold}.dim_piloto"
```

```python
from pyspark.sql import functions as F
```

#### 1 - Leer la tabla silver `pilotos`
- `silver.pilotos`


```python
drivers_df  = spark.table(f"{catalogo}.{esquema_silver}.pilotos")
#ref_nationality_region_df = spark.table(f"{catalog_name}.{gold_schema}.#ref_nationality_region")
```

#### 2 - Selecccionar las columnas requeridas:
1. pilotos.id_piloto
1. pilotos.nombre_piloto
1. pilotos.fecha_nacimiento
1. pilotos.nacionalidad
1. pilotos.region

```python
dim_drivers_df = (
    drivers_df
    
        .select(
            drivers_df.id_piloto,
            drivers_df.nombre_piloto,
            drivers_df.fecha_nacimiento,
            drivers_df.nacionalidad,
            drivers_df.region.alias("region_nacionalidad")
        )
)
```

```python
display(dim_drivers_df)
```

#### 3 - Escribir en la tabla gold `dim_piloto`

```python
(
    dim_drivers_df
        .write
        .format("delta")
        .mode("overwrite")             
        .saveAsTable(target_table)
)
```
