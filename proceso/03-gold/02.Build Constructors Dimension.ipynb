```python
target_table = f"{catalogo}.{esquema_gold}.dim_constructor"
```

```python
from pyspark.sql import functions as F
```

#### Step 1 - Read source tables
- `silver.constructors`
- `gold.ref_nationality_region`

```python
constructors_df = spark.table(f"{catalogo}.{esquema_silver}.constructores")
```

#### Step 2 - Join `constructors` with `nationality_region_df` using `nationality`
Select the following columns   
1. constructors.constructor_id 
1. constructors.constructor_name 
1. constructors.nationality 
1. ref_nationality_region.region

```python
dim_constructors_df = (
    constructors_df
        .select (
            constructors_df.id_constructor,
            constructors_df.nombre_constrcutor,
            constructors_df.nacionalidad,
            constructors_df.region.alias("region_nacionalidad")
        )
)
```

```python
display(dim_constructors_df)
```

#### Step 3 - Write the transformed data to the `gold` `dim_constructors` table

```python
(
    dim_constructors_df
        .write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(target_table)
)
```
