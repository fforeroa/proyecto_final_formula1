```python
# Define source_file and table_name
archivo_fuente = f"{ruta}/constructors.json"
nombre_tabla= f"{catalogo}.{esquema_bronze}.constructors"
```

```python
print(nombre_tabla)
```

#### Step 1 - Read the JSON file using the dataframe reader API

```python
# Define the schema
esquema_constructors = """constructorId STRING, 
                         name STRING, 
                         nationality STRING, 
                         url STRING
                         """
```

```python
# Read data from the constructors file
df_constructors = (
    spark.read
       .format('json')
       .schema(esquema_constructors)
       .option('mode', 'FAILFAST')
       .load(archivo_fuente)
)
```

```python
display(df_constructors)
```

#### Step 2 - Add Metadata Columns
- Source File
- Ingestion Timestamp

```python
df_constructors_final = add_ingesta_metadata(df_constructors)
```

#### Step 3 - Write to bronze delta table

```python
(
    df_constructors_final
        .write
        .format('delta')
        .mode('overwrite')
        .saveAsTable(nombre_tabla)
)
```
