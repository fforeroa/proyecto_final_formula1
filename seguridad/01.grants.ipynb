##  Permisos necesarios para los diferentes grupos de usuarios

### En el despliegue a producción por medio del yml se crean los siguientes grupos de usuarios:
- #### analistas_prod :Analistas de producción
- #### soporte_prod : Soporte de producción
- #### infra_prod : Infraestructura de  producción


###### Despues se asignan los siguientes permisos:

```python
%run ../../proyecto_final_formula1/PrepAmb/03.Configuracion_Ambiente
```

Permisos para realizar consultas a las tablas de producción a los analistas de datos

```python

spark.sql(f"GRANT USAGE ON SCHEMA {catalogo}.gold TO `analistas_prod` ")
spark.sql(f"GRANT SELECT ON SCHEMA {catalogo}.gold TO `analistas_prod` ")




```

Permisos para lo relacionado con infraestructura para los analistas de infraestructura

```python

%sql

GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `exlt-metastore_prod` to `infra_prod`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `exlt-bronze_prod` to `infra_prod`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `exlt-silver_prod` to `infra_prod`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `exlt-golden_prod` to `infra_prod`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `exlt-raw_prod` to `infra_prod`;
```

Permisos para soporte

```python
spark.sql(f"GRANT USAGE ON SCHEMA {catalogo}.gold TO `soporte_prod` ");
spark.sql(f"GRANT SELECT ON SCHEMA {catalogo}.gold TO `soporte_prod` ");
spark.sql(f"GRANT USAGE ON SCHEMA {catalogo}.bronze TO `soporte_prod` ");
spark.sql(f"GRANT SELECT ON SCHEMA {catalogo}.bronze TO `soporte_prod` ");
spark.sql(f"GRANT USAGE ON SCHEMA {catalogo}.silver TO `soporte_prod` ");
spark.sql(f"GRANT SELECT ON SCHEMA {catalogo}.silver TO `soporte_prod` ");
spark.sql(f"GRANT SELECT, CREATE TABLE ON SCHEMA {catalogo}.gold TO `analistas_prod` ");
```
