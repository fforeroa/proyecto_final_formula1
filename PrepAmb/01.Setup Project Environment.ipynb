# Set-up the project environment for Formula1 Project
1. Create External Location databricks-course-ext-dl1-formula1
1. Create Catalog formula1
1. Create Schemas landing, bronze, silver and gold
1. Create Volume Files in the landing schema

```sql
%python
dbutils.widgets.removeAll()
```

```sql
create widget text storageName default "datalakefforero03";
```

```sql
create widget text credential default "credential_formula1";
```

```sql
create widget text catalog default "formula1";
```

```sql
%python

storageName = dbutils.widgets.get("storageName")
catalog = dbutils.widgets.get("catalog")
credential = dbutils.widgets.get("credential")
```

### Access Cloud Storage

```sql
-- %fs ls 'abfss://formula1@datalakefforero03.dfs.core.windows.net/landing'
```

### Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt-metastore`
URL 'abfss://metastore@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `${credential}`)
COMMENT 'Ubicación externa para las tablas raw del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt-raw`
URL 'abfss://raw@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `${credential}`)
COMMENT 'Ubicación externa para las tablas raw del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt-bronze`
URL 'abfss://bronze@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `${credential}`)
COMMENT 'Ubicación externa para las tablas bronze del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt-silver`
URL 'abfss://silver@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `${credential}`)
COMMENT 'Ubicación externa para las tablas silver del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt-golden`
URL 'abfss://golden@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `${credential}`)
COMMENT 'Ubicación externa para las tablas golden del Data Lake';
```

### Create Catalog formula1

```sql
SHOW CATALOGS;
```

```sql
DROP CATALOG IF EXISTS formula1 CASCADE;
```

```sql
CREATE CATALOG IF NOT EXISTS ${catalog}
  MANAGED LOCATION 'abfss://metastore@${storageName}.dfs.core.windows.net/'
COMMENT 'Catalogo para la arquitectura medallion del ambiente de dev proyecto final';
```

### Create Schemas landing, bronze, silver, gold

```sql
DROP SCHEMA IF EXISTS formula1.raw;
DROP SCHEMA IF EXISTS formula1.bronze;
DROP SCHEMA IF EXISTS formula1.silver;
DROP SCHEMA IF EXISTS formula1.golden;
```

```sql
CREATE SCHEMA IF NOT EXISTS ${catalog}.raw;
CREATE SCHEMA IF NOT EXISTS ${catalog}.bronze
    MANAGED LOCATION 'abfss://bronze@${storageName}.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS ${catalog}.silver
    MANAGED LOCATION 'abfss://silver@${storageName}.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS ${catalog}.gold
    MANAGED LOCATION 'abfss://golden@${storageName}.dfs.core.windows.net/gold';         
```
